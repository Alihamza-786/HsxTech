```python
import asyncio
from playwright.async_api import async_playwright
from bs4 import BeautifulSoup
from utils.sitemap import get_sitemap_urls
import os

OUTPUT_DIR = "scraped_docs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

async def scrape_url(page, url):
    try:
        await page.goto(url, wait_until="networkidle", timeout=30000)
        
        # Scroll to bottom to trigger lazy loading
        await page.evaluate("window.scrollTo(0, document.body.scrollHeight)")
        await page.wait_for_timeout(2000)  # wait 2s for content to load
        
        html = await page.content()
        soup = BeautifulSoup(html, "html.parser")
        for tag in soup(["script", "style", "nav", "footer", "head"]):
            tag.decompose()
        text = soup.get_text(separator="\n", strip=True)
        return text
    except Exception as e:
        print(f"  [ERROR] {url} — {e}")
        return None

async def main():
    sitemap_urls = get_sitemap_urls("https://netsoltech.com/")
    print(f"Found {len(sitemap_urls)} URLs")

    async with async_playwright() as p:
        browser = await p.chromium.launch()
        page = await browser.new_page()

        for idx, url in enumerate(sitemap_urls, start=1):
            print(f"[{idx}/{len(sitemap_urls)}] Scraping: {url}")
            text = await scrape_url(page, url)

            if text:
                filepath = os.path.join(OUTPUT_DIR, f"{idx}.txt")
                with open(filepath, "w", encoding="utf-8") as f:
                    f.write(f"{url}\n{text}")
                print(f"  Saved: {filepath}")
            else:
                print(f"  Skipped.")

        await browser.close()

    print("Done.")

await main()
```

```python
import os
import re
from urllib.parse import urlparse

INPUT_DIR = "scraped_docs"
OUTPUT_DIR = "cleaned_docs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

STANDALONE_CTAS = {
    "read more", "download", "view", "watch now", "learn more",
    "subscribe", "email", "allow all", "customize", "allow selection",
    "deny", "consent", "details", "about", "show details",
    "pending", "yes", "no"
}

def url_to_filename(idx, url):
    parsed = urlparse(url)
    domain = parsed.netloc.replace("www.", "")
    path_parts = [p for p in parsed.path.strip("/").split("/") if p]
    if not path_parts:
        name = domain
    else:
        name = "_".join(path_parts)
    return f"{idx}.{name}.txt"

def remove_cookie_block(text):
    # Remove everything from start of cookie consent to "Deny"
    pattern = r'Consent\s*\n.*?(?:Deny|Do not sell or share my personal information)\s*\n'
    return re.sub(pattern, '', text, flags=re.DOTALL)

def is_noise_line(line):
    stripped = line.strip()

    # Empty
    if not stripped:
        return False  # handled separately

    # Placeholder patterns [#...#]
    if re.match(r'^\[#.*#\]$', stripped):
        return True

    # insights-state paths
    if stripped.startswith("insights-state-/"):
        return True

    # Cookie technical names (__cf_bm, _ga, _hjSession etc.)
    if re.match(r'^_{1,2}[a-zA-Z]', stripped) and ' ' not in stripped:
        return True

    # Duration metadata lines
    if re.match(r'^:\s*(1 day|2 years|Session|Persistent|IndexedDB|180 days|1 year|400 days|6 months|3 months|1 minute|90 days|HTML Local Storage|Pixel Tracker)', stripped, re.IGNORECASE):
        return True

    # Lines that are just numbers
    if re.match(r'^\d+$', stripped):
        return True

    # Single characters
    if len(stripped) == 1:
        return True

    # Standalone CTAs
    if stripped.lower() in STANDALONE_CTAS:
        return True

    # [empty name]
    if stripped.lower() == "[empty name]":
        return True

    # lastSelectedLanguage, lastExternalReferrer etc.
    if re.match(r'^last[A-Z]', stripped) and ' ' not in stripped:
        return True

    # tracker/pixel patterns
    if re.match(r'^[a-z]+\.(gif|png|js)$', stripped, re.IGNORECASE):
        return True

    return False

def clean_text(text):
    # Step 1: remove cookie consent block
    text = remove_cookie_block(text)

    lines = text.splitlines()
    cleaned = []
    seen_lines = set()
    prev_blank = False

    for line in lines:
        stripped = line.strip()

        # Handle blank lines
        if not stripped:
            if not prev_blank:
                cleaned.append("")
            prev_blank = True
            continue
        prev_blank = False

        # Skip noise
        if is_noise_line(stripped):
            continue

        # Skip duplicates
        if stripped in seen_lines:
            continue
        seen_lines.add(stripped)

        cleaned.append(stripped)

    return "\n".join(cleaned).strip()

# Process all files
files = sorted(os.listdir(INPUT_DIR), key=lambda x: int(x.split(".")[0]))

for filename in files:
    if not filename.endswith(".txt"):
        continue

    filepath = os.path.join(INPUT_DIR, filename)
    with open(filepath, "r", encoding="utf-8") as f:
        content = f.read()

    lines = content.splitlines()
    url = lines[0].strip()
    text = "\n".join(lines[1:])

    idx = filename.split(".")[0]
    new_filename = url_to_filename(idx, url)
    cleaned = clean_text(text)

    out_path = os.path.join(OUTPUT_DIR, new_filename)
    with open(out_path, "w", encoding="utf-8") as f:
        f.write(f"{url}\n{cleaned}")

    print(f"Saved: {new_filename}")

print("Done.")
```
