```python
import asyncio
from playwright.async_api import async_playwright
from bs4 import BeautifulSoup
from utils.sitemap import get_sitemap_urls
import os

OUTPUT_DIR = "scraped_docs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

async def scrape_url(page, url):
    try:
        await page.goto(url, wait_until="networkidle", timeout=60000)
        await page.evaluate("window.scrollTo(0, document.body.scrollHeight)")
        await page.wait_for_timeout(2000)

        html = await page.content()
        soup = BeautifulSoup(html, "html.parser")
        for tag in soup(["script", "style", "nav", "footer", "head", "iframe"]):
            tag.decompose()
        text = soup.get_text(separator="\n", strip=True)
        return text
    except Exception as e:
        print(f"  [ERROR] {url} — {e}")
        return None

async def main():
    sitemap_urls = get_sitemap_urls("https://netsoltech.com/")
    total = len(sitemap_urls)
    print(f"\nStarting scrape of {total} URLs sequentially\n")

    skipped = []

    async with async_playwright() as p:
        browser = await p.chromium.launch()
        page = await browser.new_page()

        for idx, url in enumerate(sitemap_urls, start=1):
            print(f"[{idx}/{total}] Scraping: {url}")
            text = await scrape_url(page, url)

            if text and len(text) > 100:  # skip nearly empty pages
                filepath = os.path.join(OUTPUT_DIR, f"{idx}.txt")
                with open(filepath, "w", encoding="utf-8") as f:
                    f.write(f"{url}\n{text}")
                print(f"  ✅ Saved: {idx}.txt — {len(text):,} chars")
            else:
                skipped.append((idx, url))
                print(f"  ⚠️  Skipped: {url}")

        await browser.close()

    print(f"\nDone. Saved: {total - len(skipped)} | Skipped: {len(skipped)}")
    if skipped:
        print("\nSkipped URLs:")
        for idx, url in skipped:
            print(f"  [{idx}] {url}")

await main()
```
