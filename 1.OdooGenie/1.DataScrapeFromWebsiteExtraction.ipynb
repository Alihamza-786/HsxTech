```python
import os
from urllib.parse import urlparse
import re

# Get the original URLs for better filenames
sitemap_urls = get_sitemap_urls("https://www.odoo.com/documentation/19.0/")

output_dir = "odoo_converted_docs"
os.makedirs(output_dir, exist_ok=True)

for i, doc in enumerate(docs):
    try:
        # Create filename from URL if available
        if i < len(sitemap_urls):
            url = sitemap_urls[i]
            parsed = urlparse(url)
            filename = parsed.path.strip('/').replace('/', '_')
            if not filename or filename == '':
                filename = 'homepage'
        else:
            filename = f'document_{i+1}'
        
        # Clean filename and add .md extension
        filename = re.sub(r'[^\w\-_.]', '_', filename) + '.md'
        
        # Export content
        content = doc.export_to_markdown()
        
        with open(os.path.join(output_dir, filename), 'w', encoding='utf-8') as f:
            f.write(f"# Source: {sitemap_urls[i] if i < len(sitemap_urls) else 'Unknown'}\n\n")
            f.write(content)
        
        print(f"Saved: {filename}")
        
    except Exception as e:
        print(f"Error saving document {i+1}: {e}")

print(f"\nAll documents saved to '{output_dir}' directory")
```

```python
import os
from urllib.parse import urlparse
import re

def clean_content(content: str) -> str:
    # Remove lines like comments, copyright, image placeholders, etc.
    lines = content.strip().splitlines()
    cleaned_lines = []
    for line in lines:
        # Strip unnecessary whitespace
        stripped = line.strip()
        # Skip unwanted lines
        if (
            stripped.startswith("### ") and len(stripped) < 50  # e.g., "### Odoo Standard vs..."
            or stripped.lower().startswith("copyright")
            or stripped.lower() == "comments are closed."
            or stripped.startswith("<!--")
            or re.match(r"^#+\s*One comment", stripped)
            or re.search(r"\[\.\.\.].*odoo", stripped.lower())  # very specific to your example
        ):
            continue
        cleaned_lines.append(line)
    return '\n'.join(cleaned_lines).strip()

# Get the original URLs for better filenames
sitemap_urls = get_sitemap_urls("https://hsxtech.net")
output_dir = "hsxtech_converted_docs"
os.makedirs(output_dir, exist_ok=True)

for i, doc in enumerate(docs):
    try:
        # Create filename from URL if available
        if i < len(sitemap_urls):
            url = sitemap_urls[i]
            parsed = urlparse(url)
            filename = parsed.path.strip('/').replace('/', '_')
            if not filename:
                filename = 'homepage'
        else:
            filename = f'document_{i+1}'

        # Clean filename and add .md extension
        filename = re.sub(r'[^\w\-_.]', '_', filename) + '.md'

        # Export and clean content
        raw_content = doc.export_to_markdown()
        cleaned_content = clean_content(raw_content)

        # Save to file
        with open(os.path.join(output_dir, filename), 'w', encoding='utf-8') as f:
            f.write(f"# Source: {sitemap_urls[i] if i < len(sitemap_urls) else 'Unknown'}\n\n")
            f.write(cleaned_content)

        print(f"Saved: {filename}")

    except Exception as e:
        print(f"Error saving document {i+1}: {e}")

print(f"\nAll documents saved to '{output_dir}' directory")
```
